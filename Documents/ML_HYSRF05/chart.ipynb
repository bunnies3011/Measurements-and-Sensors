```python
import matplotlib.pyplot as plt
import numpy as np

# --- Đọc dữ liệu từ file txt ---
distances = []
with open("4cm.txt", "r") as f:
    for line in f:
        if "Distance:" in line:
            parts = line.strip().split("Distance:")
            value = float(parts[1])
            distances.append(value)

# --- Tính giá trị trung bình ---
mean_distance = np.mean(distances)
true_value = 4;
# --- Vẽ đồ thị ---
plt.figure(figsize=(8, 5))
plt.plot(distances, linestyle='-', color='b', label='Khoảng cách đo được')

# Thêm đường ngang thể hiện giá trị trung bình
plt.axhline(y=mean_distance, color='r', linestyle='--', linewidth=1.5,
            label=f'Giá trị trung bình = {mean_distance:.4f} cm')

plt.axhline(y=true_value, color='black', linestyle='--', linewidth=1.5,
            label=f'True value = {4} cm')
# --- Trang trí biểu đồ ---
plt.title("Biểu đồ khoảng cách đo được")
plt.xlabel("Lần đo (index)")
plt.ylabel("Khoảng cách (cm)")
plt.grid(True, linestyle=':')
plt.legend(loc='best')

# --- Hiển thị ---
plt.show()

```

```python
import matplotlib.pyplot as plt
import numpy as np

# --- Đọc dữ liệu Distance ---
distances = []
with open("4cm.txt", "r") as f:
    for line in f:
        if "Distance:" in line:
            val = float(line.strip().split("Distance:")[1])
            distances.append(val)

# --- Tính SD và SDm ---
SD_list = []
SDm_list = []
count =1;

for i in range(1, len(distances) + 1):
    if i >= 2:  # chỉ tính SD khi có >= 2 phép đo
        current_data = distances[:i]
        SD = np.std(current_data, ddof=1)
        SDm = SD / np.sqrt(i)
    else:
        SD = 0.0
        SDm = 0.0
    SD_list.append(SD)
    SDm_list.append(SDm)
    count = count +1
    if (count == 2000):
        break;



# --- Tính trung bình ---
mean_SD = np.mean(SD_list[1:])   # bỏ phần đầu có thể là 0
mean_SDm = np.mean(SDm_list[1:])

# --- Vẽ đồ thị ---
plt.figure(figsize=(9, 5))
plt.plot(SD_list, marker='o', linestyle='-', color='b', label='SD (Độ lệch chuẩn từng lần đo)')
plt.plot(SDm_list, marker='s', linestyle='--', color='orange', label='SDm = SD / √n')

plt.axhline(y=mean_SD, color='b', linestyle=':', linewidth=1.5, label=f'Trung bình SD = {mean_SD:.4f}')
plt.axhline(y=mean_SDm, color='orange', linestyle=':', linewidth=1.5, label=f'Trung bình SDm = {mean_SDm:.4f}')

plt.title("Sự thay đổi của SD và SDm theo số lần đo")
plt.xlabel("Số lần đo (index)")
plt.ylabel("Giá trị SD / SDm")
plt.grid(True, linestyle=':')
plt.legend(loc='best')
plt.tight_layout()
plt.show()
```
